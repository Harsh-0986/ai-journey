```python
import numpy as np
```

# Vector Addition

```python
def add(v1, v2):
    v3 = np.empty(v1.shape)
    for i in range(v1.size):
        v3[i] = v1[i] + v2[i]

    return v3
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

add(a, b) == (a + b)
```

# Dot Product

```python
def dot_product(v1, v2):
    ans = 0

    for i in range(v1.size):
        ans += v1[i] * v2[i]

    return ans
    
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product(a, b) == np.dot(a,b)
```

# Matrix Multiplication

```python
def matrix_multiplication(A, B):
    if A.shape[1] != B.shape[0]: raise Exception("Matrix shapes are invalid")
    C = np.empty([A.shape[0], B.shape[1]])

    for row in range(A.shape[0]):
        for col in range (B.shape[1]):
            C[row][col] = dot_product(A[row], B[:, col])

    return C
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

matrix_multiplication(A, B) == np.matmul(A, B)
```
